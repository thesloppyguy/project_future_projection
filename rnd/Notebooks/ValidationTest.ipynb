```python
import pandas as pd
```

```python
df = pd.read_excel('../data/Sales Jul.xlsx', sheet_name='Jul25 - Oct25')
```

```python
df.head()
```

```python
df_subset = df[['Material', 'SALES OFFICE CODE', 'Segment','Star Rating.1', 'Billing Date', 'Billing Quantity ODU','Billing Quantity IDU', 'Tonnage']]
```

```python
df_subset.fillna(0, inplace=True)
df_subset = df_subset[df_subset['Billing Quantity ODU'] > 0]
df_subset.rename(columns={
    'Billing Date': 'Date', 
'Material': 'Item Code', 
'SALES OFFICE CODE': 'Branch', 'Star Rating.1': 'Star Rating', 'Billing Quantity ODU': 'Quantity', 'Tonnage': 'Tonnage'}, inplace=True)
df_subset['Date'] = pd.to_datetime(df_subset['Date'])
df_subset.drop(columns=['Billing Quantity IDU'], inplace=True)
df_subset['Region'] = 'South'
```

```python
df_subset
```

```python
main_df = pd.read_csv('../data/merged_filter_ingestion.csv')
main_df['Date'] = pd.to_datetime(main_df['Date'])
main_df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
merged_df = pd.concat([main_df, df_subset])
```

```python
merged_df.to_csv('../data/merged_filter_ingestion_latest.csv', index=False)
```
