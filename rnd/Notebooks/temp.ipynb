```python
fs_df = pd.read_csv('../data/Final Sales.csv')
```

```python
fs_df
```

```python
# Add 'Financial Year' column and compute yearly sales by financial year (April-March)
fs_df['Date'] = pd.to_datetime(fs_df['Date'])

def get_fin_year(dt):
    year = dt.year
    if dt.month >= 4:
        return f"{year}-{year+1}"
    else:
        return f"{year-1}-{year}"

fs_df['Financial Year'] = fs_df['Date'].apply(get_fin_year)
fin_yearly_sales = fs_df.groupby('Financial Year')['Sales Qty.'].sum()
fin_yearly_sales

```
